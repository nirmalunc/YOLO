Import Libraries

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from ultralytics.utils.loss import v8DetectionLoss
from ultralytics import YOLO
```

```python
# Fixed seed so we can ensure losses are different based on diff. loss functions and not randomness
torch.manual_seed(42)

# Models with different loss functions
model_baseline = YOLO('yolov8n.pt')
# model_baseline_boxdiff = create_model_with_custom_loss(model_type='yolov8n.pt', loss_type='baseline', box=10)
# model_vsr = create_model_with_custom_loss(model_type='yolov8n.pt', loss_type='vsr', vsr_weight=0.2)
# model_sbce = create_model_with_custom_loss(model_type='yolov8n.pt', loss_type='sbce', sbce_max_size=0.05)
# model_both = create_model_with_custom_loss(model_type='yolov8n.pt', loss_type='both', vsr_weight=0.2, sbce_max_size=0.05)

# Training hyperparameters
train_config = {
    'data': 'coco128.yaml',
    'epochs': 10,            # Reasonable number for demo
    'imgsz': 416,            # Image size
    'batch': 16,             # Batch size
    'patience': 5            # Early stopping patience
}

def train_model(model, name):
    return model.train(
        **train_config,
        name=name
    )
```

```python
results_baseline = train_model(model_baseline, 'baseline')
```
